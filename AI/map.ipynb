```python
import cv2
import numpy as np

# Load the image
image = cv2.imread("mapout12.png")

# Convert to HSV color space
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# Define the lower and upper range for red color
lower_red1 = np.array([0, 70, 50])
upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([170, 70, 50])
upper_red2 = np.array([180, 255, 255])

# Create masks for both red ranges
mask1 = cv2.inRange(hsv, lower_red1, upper_red1)
mask2 = cv2.inRange(hsv, lower_red2, upper_red2)

# Combine both masks
mask = cv2.bitwise_or(mask1, mask2)

# Create a black background (all zeros)
output = np.zeros_like(image)

# Set the red lines (from mask) to white (255,255,255)
output[mask > 0] = [255, 255, 255]

# Convert to grayscale
gray_output = cv2.cvtColor(output, cv2.COLOR_BGR2GRAY)

# Save or display the result
cv2.imwrite("black_white_result.pgm", gray_output)
#cv2.imshow("Result", gray_output)
#cv2.waitKey(0)
#cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Load the image you prepared
image = cv2.imread("mapout12.png")

# Convert the image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Invert colors if necessary (to match ROS occupancy map conventions)
_, binary_map = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY_INV)

# Save the map as a .pgm (portable graymap)
cv2.imwrite("mapros.png", binary_map)
print("Map saved as map.pgm")
```
